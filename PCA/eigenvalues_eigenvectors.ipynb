#### Extracting Eigenvalues and eigenvectors

```python
import numpy as np
```

```python
# 3x3 matrix
A_3x3 = np.array([
    [6, 2, 1],
    [2, 3, 1],
    [1, 1, 1]
])

# 4x4 matrix
A_4x4 = np.array([
    [5, 2, 1, 1],
    [2, 6, 2, 1],
    [1, 2, 7, 3],
    [1, 1, 3, 4]
])
```

```python
# Compute eigenvalues and eigenvectors for each
eigvals_3x3, eigvecs_3x3 = np.linalg.eig(A_3x3)
eigvals_4x4, eigvecs_4x4 = np.linalg.eig(A_4x4)

results_eg_vs = {
    "3x3 Eigenvalues": eigvals_3x3,
    "3x3 Eigenvectors": eigvecs_3x3,
    "4x4 Eigenvalues": eigvals_4x4,
    "4x4 Eigenvectors": eigvecs_4x4
}
```

```python
print(results_eg_vs)
```

```python
from sympy import symbols, Eq, solve

x = symbols('x')
eq = x**3 - 10*x**2 + 22*x - 15
roots = solve(eq)
roots_numeric = [r.evalf() for r in roots]
print(roots_numeric)
```
