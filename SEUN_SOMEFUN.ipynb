<h3 style="text-align:center">EUROPEAN CALL OPTION</h3>



---

A **European call option** gives the holder the right (but not the obligation) to **buy** an underlying asset at a specified **strike price $E$** at a specified **maturity time $T$**. It can only be exercised at **maturity**, unlike American options.

---

## Notation and Symbols

| Symbol         | Meaning                                  |
|----------------|-------------------------------------------|
| $ S_t $      | Underlying asset price at time $ t $   |
| $ E $        | Strike price                             |
| $ T $        | Time of maturity                         |
| $ r $        | Risk-free interest rate (annualized)     |
| $ \sigma\ $    | Volatility of the asset (standard deviation) |
| ${E}^{\mathbb{Q}} $ | Expectation under risk-neutral measure |
| $V(S, t)$  | Value of the option at time $t$ when asset price is $S$ |

---

## Payoff at Maturity

At maturity $ T $, the value of a European call option is:

$$
\text{Payoff} = \max(S_T - E, 0)
$$

---

## Risk-Neutral Valuation

Under the **risk-neutral measure ${Q} $**, the value of the call option at time \( t \) is:

$$
V(S_t, t) = e^{-r(T - t)} \mathbb{E}^{\mathbb{Q}} \left[ \max(S_T - E, 0) \mid S_t \right]
$$

This is the **discounted expected payoff** under the risk-neutral probability measure.

---

## Black-Scholes Formula for European Call

The closed-form solution (Black-Scholes-Merton model) for a European call is:

$$
V(S_t, t) = S_t \N(d_1) - E e^{-r(T - t)} \N(d_2)
$$

Where:

$$
d_1 = \frac{\ln(S_t / E) + (r + \frac{1}{2} \sigma^2)(T - t)}{\sigma \sqrt{T - t}} \quad and \quad
d_2 = d_1 - \sigma \sqrt{T - t},

\quad

N(x) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^{x} e^{-\frac{1}{2} \phi^2} \, d\phi


$$

---


## Practical Insights

- **In-the-money**: $ S > E $
- **Out-of-the-money**: $ S < E $
- **At-the-money**: $ S \approx E $

European options are widely used in theoretical finance because of their closed-form solutions and easier risk analysis.

<h3 style="text-align:center">BINARY CALL OPTION</h3>

Binary options, also known as **digital options**, are **zero-one bets** on the underlying asset’s performance. There are two general types:

- **Cash-or-nothing**
- **Asset-or-nothing**

---

## Cash-or-Nothing vs Asset-or-Nothing

- **Cash-or-nothing:** Pays a fixed amount of cash (typically 1 unit) at expiry **if the option is in-the-money**.
- **Asset-or-nothing:** Pays out the **value of the underlying asset** at expiry if in-the-money.

---

## Heaviside "Unit Step" Function

The payoff is defined in terms of the **Heaviside step function** $H(x) $, which is commonly represented as:

$$
H(x) = 
\begin{cases}
1 & \text{if } x > 0 \\
0 & \text{if } x < 0
\end{cases}
$$

Alternative definitions include:

$$
H(x) = 
\begin{cases}
1 & x > 0 \\
\frac{1}{2} & x = 0 \\
0 & x < 0
\end{cases}
\quad \text{or} \quad
H(x) = 
\begin{cases}
1 & x \geq 0 \\
0 & x < 0
\end{cases}
$$

> Normalising the payoff to one (i.e., £1-or-nothing) is often done to facilitate mathematical modelling. In practice, it is often set to a fixed amount like £100.

---

## Binary Call Option Payoff

For a binary call option:

$$
C_b(S, T) = H(S(T) - E) = 
\begin{cases}
1 & \text{if } S(T) \geq E \\
0 & \text{otherwise}
\end{cases}
$$

Here:
- $S(T)$ is the asset price at maturity,
- $E$ is the strike price.

---

## Binary Call Option Value

$$
C_b(S, t) = e^{-r(T - t)} N(d_2)
$$

Where:
- $r$: Risk-free interest rate
- $Nd_2$: CDF of the standard normal distribution
- $d_2$: Usual Black-Scholes definition:
  
$$
d_2 = \frac{\ln(S / E) + (r - \frac{1}{2}\sigma^2)(T - t)}{\sigma \sqrt{T - t}}
$$

---

## Binary Put Option Payoff

$$
P_b(S, T) = H(E - S(T)) = 
\begin{cases}
1 & \text{if } S(T) \leq E \\
0 & \text{otherwise}
\end{cases}
$$

## Binary Put Option Value

$$
P_b(S, t) = e^{-r(T - t)} N(-d_2)
$$

---

## Pricing Formulas

### Cash-or-Nothing (Example for FX where $ r_d $ is domestic interest rate)

- **Call / Domestic Put:**

$$
C = e^{-r_d \tau} N(d_2)
$$

- **Put / Domestic Call:**

$$
P = e^{-r_d \tau} N(-d_2)
$$

---

### Asset-or-Nothing (Example for FX where $r_f$ is foreign interest rate)

- **Call / Domestic Put:**

$$
C = S e^{-r_f \tau} N(d_1)
$$

- **Put / Domestic Call:**

$$
P = S e^{-r_f \tau} N(-d_1)
$$

---

## Put-Call Parity for Digital Options

$$
C + P = e^{-r_d \tau} N(d_2) + e^{-r_d \tau} N(-d_2) = e^{-r_d \tau}
$$

This means the total value of a binary call and binary put (on the same underlying and strike) equals the **discounted value of a unit of cash**.

<h3 style="text-align:center">NUMERICAL METHODS FOR SDEs</h3>

### Numerical Methods for Stochastic Differential Equations (SDEs)

In quantitative finance and stochastic modeling, numerical methods are essential for simulating stochastic differential equations (SDEs) that often lack closed-form solutions. Two prominent schemes are:

- **Euler-Maruyama Scheme**
- **Milstein Scheme**

These methods are widely used for simulating asset price paths and other stochastic processes.

---

## Euler-Maruyama Scheme

The Euler-Maruyama method is an extension of the Euler method for ODEs to stochastic settings. It approximates SDE solutions via time discretization and simulates the process incrementally.

### Mathematical Formulation

Given the SDE:

$$
dX_t = \mu(X_t, t)\, dt + \sigma(X_t, t)\, dW_t
$$

The Euler-Maruyama discretization is:

$$
X_{t + \Delta t} = X_t + \mu(X_t, t)\, \Delta t + \sigma(X_t, t)\, \Delta W_t
$$

where $\Delta W_t \sim \mathcal{N}(0, \Delta t)$ is a Wiener increment.

### Characteristics

- **Strong Convergence Order:** ${O}(\sqrt{\Delta t})$
- **Advantages:** Easy to implement, fast computation
- **Limitations:** Less accurate for nonlinear diffusion; requires small time steps for precision



Euler-Maruyama is simple and sufficient when coefficients are constant. However, when the diffusion term is state-dependent, its approximation becomes less reliable.

---

## Milstein Scheme

The Milstein method improves upon Euler-Maruyama by including a correction term based on the derivative of the diffusion coefficient. This makes it more accurate for nonlinear diffusion functions.

### Mathematical Formulation

$$
X_{t + \Delta t} = X_t + \mu(X_t, t)\, \Delta t + \sigma(X_t, t)\, \Delta W_t + \frac{1}{2} \sigma(X_t, t) \frac{\partial \sigma(X_t, t)}{\partial X} \left[ (\Delta W_t)^2 - \Delta t \right]
$$

### Characteristics

- **Strong Convergence Order:** $\mathcal{O}(\Delta t)$
- **Advantages:** More accurate for state-dependent volatility
- **Limitations:** Requires derivative ${\partial \sigma}/{\partial X}$; slightly more complex



The Milstein method is particularly valuable when simulating SDEs with stochastic or nonlinear volatility. The additional correction term enhances fidelity by accounting for curvature due to the Brownian motion.

---

## Summary Comparison

| Feature                 | Euler-Maruyama                   | Milstein                          |
|-------------------------|----------------------------------|-----------------------------------|
| **Strong Convergence**  | ${O}(\sqrt{\Delta t})$    |  ${O}(\Delta t)$        |
| **Implementation**      | Simple                           | Moderate (requires $\sigma'$)     |
| **Best For**            | Constant volatility              | Nonlinear or stochastic volatility|
| **Correction Term**     | None                             |     Included                       |

### Black Scholes Model

The *Black Scholes* model which is an example of a closed form solution and the one to be considered in this miniproject was published in 1973 for non-dividend paying stocks and since then the model has created a revolution in quantitative finance and opened up derivatives pricing paradigm. Black Scholes model is based on number of assumptions about how financial markets operate and those are:

* Arbitrage Free Markets
* Frictionless and Continuous Markets
* Risk Free Rates
* Log-normally Distributed Price Movements
* Constant Volatility

These assumptions maynot hold true in reality, but are not particularly limiting. The generalized Black Scholes framework have been extended to price derivaties of other asset classes such as Black 76 (Commodity Futures) and Garman-Kohlhagen (FX Futures) that are currently used in derivative pricing and risk management.


### Black Scholes Formula
The Black–Scholes equation describes the price of the option over time as

$$
\frac{\partial V}{\partial t}
+ \frac 1{2}{\sigma^2 S^2} \frac{\partial^2 V}{\partial S^2}
+ r S \frac{\partial V}{\partial S} - rV = 0 
$$

<br>Solving the above equation, we know that the value of a call option for a non-dividend paying stock is:<br>

$$ C = SN(d_1) - Ke^{-rt}N(d_2) $$

and, the corresponding put option price is:

$$ P = Ke^{-rt}N(-d_2) - SN(-d_1)$$

where, 

$$ d_1= \frac{1}{\sigma \sqrt{t}}\left[\ln{\left(\frac{S}{K}\right)} +{\left(r + \frac{\sigma^2}{2}\right)}t\right] $$
<br>

$$ d_2= d_1 - \sigma \sqrt{t} $$
<br>

$$ N(x)=\frac{1}{\sqrt{2\pi}} \int_{-\infty}^{x} \mathrm e^{-\frac{1}{2}x^2} dx $$


$S$ is the spot price of the underlying asset<br>
$K$ is the strike price<br>
$r$ is the annualized continuous compounded risk free rate<br>
$\sigma$ is the volatility of returns of the underlying asset<br>

### Option Pricing Techniques 

As with other option pricing techniques Monte Carlo methods are used to price options using what is essentially a three step process.

**Step 1:** Simulate potential price paths of the underlying asset.<br>
**Step 2:** Calculate the option payoff for each of these price paths.<br>
**Step 3:** Average the payoff and discount back to today to determine the option price.

<br>

### Simulating Asset Prices

Next, we will simulate the asset price at maturity $S_{T}$. Following Black-Scholes-Merton where the underlying follows under risk neutrality, a geometric Brownian motion with a stochastic differential equation (SDE) is given as


\begin{equation*}
   dS_{t} = rS_{t}dt + σS_{t}dW_{t} 
\end{equation*}

where $S_{t}$ is the price of the underlying at time $t$, σ is constant volatility, $r$ is the constant risk-free interest rate and $W$ is the brownian motion.<br>

Applying Euler discretization of SDE, we get

\begin{equation*}
   S_{t+{\delta}t} = S_t * (1 + {r {\delta}t + {\sigma} {\sqrt{\delta}t} w_{t})}
\end{equation*}

It is often more convenient to express in time stepping form

\begin{equation*}
   S_{t+{\delta}t} = S_t exp^{((r-\frac{1}2{\sigma}^2){\delta}t + {\sigma} {\sqrt{\delta}t} w_{t})}
\end{equation*}


The variable w is a standard normally distributed random variable, 0 < ${\delta}t$ < T, time interval. It also holds 0 < t ≤ T with T the final time horizon.

### CLOSED FORM SOLUTION (BLACK SCHOLES)

```python
# Black-Scholes closed-form option payoff calculator

def black_scholes_call_price(spot, strike, rate, dte, volatility, option='european'):
    """
    Calculate the Black-Scholes price for a European or Binary call option.

    Parameters:
    - spot: Current price of the underlying asset (S)
    - strike: Strike price of the option (E)
    - rate: Risk-free interest rate (r)
    - dte: Time to expiration in years (t)
    - volatility: Volatility of the underlying asset (sigma)
    - option: 'european' or 'binary'

    Returns:
    - float containing the call option price
    """
    if option == 'european':
        time_sqrt = np.sqrt(dte)
        discount_factor = np.exp(-rate * dte)

        d1 = (np.log(spot / strike) + (rate + 0.5 * volatility ** 2) * dte) / (volatility * time_sqrt)
        d2 = d1 - volatility * time_sqrt

        return spot * norm.cdf(d1) - strike * discount_factor * norm.cdf(d2)

    elif option == 'binary':
        d2 = (np.log(spot / strike) + (rate - 0.5 * volatility ** 2) * dte) / (volatility * np.sqrt(dte))
        return np.exp(-rate * dte) * norm.cdf(d2)

    else:
        raise ValueError("Invalid option type. Choose 'european' or 'binary'.")

```

### NUMERICAL METHODS (EULER AND MILSTEIN SCHEMES)

#### EULER-MARUYAMA SCHEME

```python
# Parameters
spot = 100
mu = 0.05
sigma = 0.2
horizon = 1.0
timesteps = 252
nsims = 100000
strike = 100
r = 0.05


# Generate path
def simulate_path_euler(spot, mu, sigma, horizon, timesteps, nsims):
    
    # set the seed
    np.random.seed(2025)

    # define dt
    dt = horizon/timesteps
    
    # simulate path
    S = np.zeros((timesteps,nsims))
    S[0] = spot
    
    for i in range(0, timesteps-1):
        w = np.random.standard_normal(nsims)
        S[i+1] = S[i] * (1+ mu*dt + sigma*np.sqrt(dt)*w)
    
    return S
```

```python
# Calculate the discounted value of the expeced payoff for European and binary option
european_payoff_euler = np.exp(-r*T) * np.mean(np.maximum(0, S[-1]-K))
binary_payoff_euler = np.exp(-r * T) * np.mean((S[-1] > K).astype(float))


# Print the values
print(f"European Call Option Value is {european_payoff_euler:0.4f}")
print(f"binary Call Option Value is {binary_payoff_euler:0.4f}")
```

#### MILSTEIN SCHEME

```python
# Milstein simulation function
def simulate_path_milstein(spot, mu, sigma, horizon, timesteps, nsims):
    np.random.seed(2025)
    dt = horizon / timesteps
    S = np.zeros((timesteps, nsims))
    S[0] = spot
    
    for i in range(0, timesteps-1):
        z = np.random.standard_normal(nsims)
        S[i+1] = S[i] * (1 + mu*dt + sigma*np.sqrt(dt)*z + 0.5 * sigma**2 * dt * (z**2 - 1))
    
    return S
```

### STANDARD ERROR ANALYSIS

```python
# Number of simulations
n = 100000

# Define parameters
K = 100; r = 0.05; T=1

# --- European Call ---

S = simulate_path_euler(100,0.05,0.2,1,252,100000)

P = simulate_path_milstein(100,0.05,0.2,1,252,100000)

euro_payoffs_euler = np.exp(-r*T) * np.maximum(0, S[-1] - K)
euro_se_euler = np.std(euro_payoffs_euler, ddof=1) / np.sqrt(n)

euro_payoffs_milstein = np.exp(-r*T) * np.maximum(0, P[-1] - K)
euro_se_milstein = np.std(euro_payoffs_milstein, ddof=1) / np.sqrt(n)

# --- Binary Call ---
binary_payoffs_euler = np.exp(-r*T) * (S[-1] > K).astype(float)
binary_se_euler = np.std(binary_payoffs_euler, ddof=1) / np.sqrt(n)

binary_payoffs_milstein = np.exp(-r*T) * (P[-1] > K).astype(float)
binary_se_milstein = np.std(binary_payoffs_milstein, ddof=1) / np.sqrt(n)

# Print results
print(f"Standard Errors:")
print(f"European Call (Euler):    {euro_se_euler:.6f}")
print(f"European Call (Milstein): {euro_se_milstein:.6f}")
print(f"Binary Call (Euler):      {binary_se_euler:.6f}")
print(f"Binary Call (Milstein):   {binary_se_milstein:.6f}")
```

#### Standard Error (SE) in Monte Carlo Simulations

<b>Standard Error (SE)</b> measures the statistical uncertainty or volatility in an estimate due to randomness in the simulation.

- A <b>smaller SE</b> means the estimate is more stable and closer to the true value.
- A <b>larger SE</b> indicates more noise, meaning the estimate might vary more with repeated simulations.



#### Interpretation of European Call Option


- Monte Carlo estimate ≈ **10.42**
- Standard Error (SE) ≈ **0.0463**

This tells us that the Monte Carlo estimate of the European call option price (around **10.42**) has an uncertainty of **±0.0463**.



#### 95% Confidence Interval

With **95% confidence** (± 1.96 × SE), the true value lies in:

$$
10.42 \pm (1.96 \times 0.0463)
$$

Which evaluates to:

$$
[10.33,   10.51]
$$


#### Interpretation of binary Call Option

- Monte Carlo estimate ≈ **0.533**
- Standard Error $(SE)$ ≈ **0.001493**

The value of the binary call is much smaller(around 0.533) which means it is very precise.

With **95% confidence** (± 1.96 × SE), the true value lies in:

$$
0.533 \pm (1.96 \times 0.001493)
$$

Which evaluates to:

$$
[0.530,   0.536]
$$

#### Summary of SE Result

The summary of this result is that **binary options** have a payoff of **0 or 1**, so the distribution of simulated outcomes is **less spread out** (lower Standard Error, SE).

In contrast, **European options** have a **wider range of possible payoffs**, especially as **volatility increases**, which leads to a **higher Standard Error (SE)**.

This means that:

- **Binary options** produce more stable Monte Carlo estimates.
- **European options** introduce more variability in their estimates as they can take on a continuous range of payoffs depending on the final asset price.

<h2 style="text-align:center; font-weight:bold;">VARYING NUMBER OF SIMULATIONS</h2>

```python
# Vary number of simulations from 10k to 500k
sim_counts = np.arange(10000, 510000, 50000)

# Parameters
spot = 100
mu = 0.05
sigma = 0.2
r = 0.05
T = 1
timesteps = 252
K = 100

# Collect results
results_sims = []

for nsims in sim_counts:
    # Simulate paths
    S_euler = simulate_path_euler(spot, mu, sigma, T, timesteps, nsims)
    S_milstein = simulate_path_milstein(spot, mu, sigma, T, timesteps, nsims)
    
    # Black-Scholes
    bs_ec = black_scholes_call_price(spot, K, r, T, sigma, 'european')
    bs_bc = black_scholes_call_price(spot, K, r, T, sigma, 'binary')
    
    # Euler-Maruyama payoffs
    euler_ec = np.exp(-r*T) * np.mean(np.maximum(0, S_euler[-1] - K))
    euler_bc = np.exp(-r*T) * np.mean((S_euler[-1] > K).astype(float))
    
    # Milstein payoffs
    milstein_ec = np.exp(-r*T) * np.mean(np.maximum(0, S_milstein[-1] - K))
    milstein_bc = np.exp(-r*T) * np.mean((S_milstein[-1] > K).astype(float))
    
    # Errors
    euler_ec_error = abs(euler_ec - bs_ec)
    milstein_ec_error = abs(milstein_ec - bs_ec)
    euler_bc_error = abs(euler_bc - bs_bc)
    milstein_bc_error = abs(milstein_bc - bs_bc)
    
    results_sims.append([nsims, euler_ec, euler_ec_error, milstein_ec, milstein_ec_error,
                         euler_bc, euler_bc_error, milstein_bc, milstein_bc_error])

# Table of results
headers = ['N sims', 'Euler Euro', 'Error', 'Milstein Euro', 'Error',
           'Euler Binary', 'Error', 'Milstein Binary', 'Error']

print(tabulate(results_sims, headers=headers, floatfmt=".4f", tablefmt="fancy_outline"))
```
